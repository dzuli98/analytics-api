```python
import requests
import pprint
```

```python
path = '/api/events'
base_url = 'http://localhost:8000'
endpoint = f'{base_url}{path}'
```

```python
response = requests.get(endpoint, params={'duration': '1 minute', 'pages': []})
if response.ok:
    print("API is reachable.")
    data = response.json()
    pprint.pprint(data)
```
