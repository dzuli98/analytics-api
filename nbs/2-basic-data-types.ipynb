```python
import requests
```

```python
path = '/api/events/10'
base_url = 'http://localhost:8000'
endpoint = f'{base_url}{path}'
```

```python
response = requests.get(endpoint)
if response.ok:
    print("API is reachable.")
    data = response.json()
    print("Response JSON:", data, type(data))
```

```python
headers = {
    'Content-Type': 'application/json'
}

create_endpoint = f'{base_url}/api/events'
response = requests.post(create_endpoint, json={'page':'test+'}, headers=headers)
print(response.headers)
print(response.ok)
if response.ok:
    print(response.json())
else:
    print(response.text)
```

```python
headers = {
    'Content-Type': 'application/json'
}

update_endpoint = f'{base_url}/api/events/2'
response = requests.put(update_endpoint, json={'description': 'NEW!'}, headers=headers)
print(response.headers)
print(response.ok)
if response.ok:
    print(response.json())
else:
    print(response.text)
```

```python
from pprint import pprint
```

```python
path = '/api/events/10'
base_url = 'http://localhost:8000'
endpoint = f'{base_url}{path}'
response = requests.get(endpoint)
if response.ok:
    pprint(response.json())
else:
    print("Status:", response.status_code)
    print("Error:", response.text)
```

```python
import random  
event_count = 1000

pages = ['/about', '/contact', '/pages', '/pricing']

for i in range(event_count):
    page = random.choice(pages)
    response = requests.post(create_endpoint, json={'page': page})
    if response.ok:
        data = response.json()
        print(f"Created event {data['id']} for page {data['page']}")
    else:
        print("Error:", response.text)
```
