```python
import requests
```

```python
path = '/api/events/10'
base_url = 'http://localhost:8000'
endpoint = f'{base_url}{path}'
```

```python
response = requests.get(endpoint)
if response.ok:
    print("API is reachable.")
    data = response.json()
    print("Response JSON:", data, type(data))
```

```python
headers = {
    'Content-Type': 'application/json'
}

create_endpoint = f'{base_url}/api/events'
response = requests.post(create_endpoint, json={'page':'test+'}, headers=headers)
print(response.headers)
print(response.ok)
if response.ok:
    print(response.json())
else:
    print(response.text)
```

```python
headers = {
    'Content-Type': 'application/json'
}

update_endpoint = f'{base_url}/api/events/10'
response = requests.put(update_endpoint, json={'description': 'hello world'}, headers=headers)
print(response.headers)
print(response.ok)
if response.ok:
    print(response.json())
else:
    print(response.text)
```
